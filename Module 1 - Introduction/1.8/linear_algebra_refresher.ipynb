## Multiplication

```python
# vector vector multiplication

v = np.array([1,1,2,0])

```

```python
def vec_vec_mul(u,v):
    assert u.shape[0] == v.shape[0]
    
    n = u.shape[0]
    
    result=0.0
    
    for i in range(n):
        result = result + u[i]*v[i]
        
    return result
```

```python
vec_vec_mul(u,v)
```

```python
u.dot(v)
```

```python
# Matrix vector multiplication

U = np.array([
    [2,4,5,6],
    [1,2,1,2],
    [3,1,2,1]
])
```

```python
def mat_vec_mul(U,v):
    assert U.shape[1]==v.shape[0]
    
    num_rows=U.shape[0]
    
    result = np.zeros(num_rows)
    
    for i in range(num_rows):
        result[i] = vec_vec_mul(U[i], v)
        
    return result
```

```python
mat_vec_mul(U,v)
```

```python
#Matrix Matrix multiplication


V = np.array([
    [1,1,2],
    [0,0.5,1],
    [0,2,1],
    [2,1,0]
]) 
```

```python
def mat_mat_mul(U,V):
    assert U.shape[1]==V.shape[0]
    
    num_rows = U.shape[0]
    num_cols = V.shape[1]
    
    result = np.zeros((num_rows, num_cols))
    
    for i in range(num_cols):
        vi = V[:, i]
        Uvi = mat_vec_mul(U, vi)
        result[: ,i] = Uvi
        
    return result
```

```python
mat_mat_mul(U,V)
```
